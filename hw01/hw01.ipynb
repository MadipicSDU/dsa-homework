````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
cities_clean = [city.strip().lower().capitalize() for city in cities]
amounts_int = [int(amount) for amount in amounts]
check_number = len(amounts_int)
total_sum = sum(amounts_int)
average_amount = round(total_sum / check_number)
unpaid_checks = sum(1 for p in paid if not p)
unpaid_sum = sum(amount for amount, p in zip(amounts_int, paid) if not p)
print(f"Количество заказов: {check_number}")
print(f"Общая сумма: {total_sum}")
print(f"Средний чек: {average_amount}")
print(f"Неоплаченные заказы: {unpaid_checks}, сумма: {unpaid_sum}")
for city, amount, status in zip(cities_clean, amounts_int, paid):
    print(f"{city} | {amount} | {'Оплачен' if status else 'Не оплачен'}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
revenue_almaty = sum(amount for amount, p, city in zip(amounts_int, paid, cities_clean) if p and city == "Almaty")
revenue_astana = sum(amount for amount, p, city in zip(amounts_int, paid, cities_clean) if p and city == "Astana")
print(f"Выручка Алматы: {revenue_almaty}")
print(f"Выручка Астаны: {revenue_astana}")
largest_paid_index = max((i for i, p in enumerate(paid) if p), key=lambda i: amounts_int[i])
largest_paid_city = cities_clean[largest_paid_index]
print(f"Самый крупный оплаченный заказ: индекс {largest_paid_index}, {largest_paid_city}, {amounts_int[largest_paid_index]}")
weekend_order_ratio = sum(1 for day in days if day in ["сб", "вс"]) / check_number
print(f"Доля заказов за выходные: {weekend_order_ratio:.2%}")
i = 0
savings_fixed = 0
while savings_fixed < 400000:
    savings_fixed += 15000
    i += 1
j = 0
savings_increasing = 0
monthly_contribution = 15000
while savings_increasing < 400000:
    savings_increasing += monthly_contribution
    monthly_contribution += 1000
    j += 1
print(f"Накопление по 15000 в месяц: {i} месяцев")
print(f"Накопление с увеличением взноса: {j} месяцев")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
unique_cities = list(set(cities_clean))
revenue_by_city = [0] * len(unique_cities)
for city, amount, p in zip(cities_clean, amounts_int, paid):
    if p:
        index = unique_cities.index(city)
        revenue_by_city[index] += amount
for i in range(len(unique_cities)):
    for j in range(i + 1, len(unique_cities)):
        if revenue_by_city[i] < revenue_by_city[j]:
            revenue_by_city[i], revenue_by_city[j] = revenue_by_city[j], revenue_by_city[i]
            unique_cities[i], unique_cities[j] = unique_cities[j], unique_cities[i]
for city, revenue in zip(unique_cities, revenue_by_city):
    print(f"{city} | {revenue}")
amounts_sorted = sorted(amounts_int)
middle = len(amounts_sorted) // 2
if len(amounts_sorted) % 2:
    median_revenue = amounts_sorted[middle]
else:
    median_revenue = (amounts_sorted[middle - 1] + amounts_sorted[middle]) / 2
print(f"Медиана суммы заказа: {median_revenue:g}")
week_days = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]
day_revenue = []
for day in week_days:
    revenue = sum(amount for amount, order_day, p in zip(amounts_int, days, paid) if p and order_day == day)
    day_revenue.append(revenue)
    print(f"{day} | {'█' * (revenue // 5000)}  {revenue}")
````

В исходных названиях городов встречаются лишние пробелы и разный регистр, поэтому один город может выглядеть как несколько разных значений. Удаление пробелов по краям и единый регистр позволяют правильно группировать заказы. Суммы записаны строками, поэтому перед подсчётом общей суммы, среднего чека и медианы их нужно преобразовать в числа.
